## Cross-Dataset Relationship Check

### Objective

Memverifikasi bahwa hubungan antar dataset konsisten berdasarkan key yang dimiliki masing-masing tabel.

### Output

- Validasi hubungan antar tabel.
- Jumlah foreign key yang tidak memiliki pasangan (jika ada).
- Status setiap relationship (PASSED / FAILED).

### Notes

Tahap ini hanya melakukan pengecekan keberadaan foreign key dan belum melakukan validasi bisnis secara mendalam.

### Customers → Orders

**Objective**

Memastikan setiap `customer_id` pada tabel **orders** memiliki pasangan pada tabel **customers**.

```python
missing_customers = orders.loc[
    ~orders["customer_id"].isin(customers["customer_id"])
]

print("=" * 70)
print("Relationship Check : Customers → Orders")
print("=" * 70)

print(f"Total Orders                 : {len(orders):,}")
print(f"Orders without Customer      : {len(missing_customers):,}")

if len(missing_customers) == 0:
    print("Relationship Status          : PASSED ✅")
else:
    print("Relationship Status          : FAILED ❌")
```

### Orders → Order Items

**Objective**

Memastikan seluruh `order_id` pada tabel **order_items** memiliki pasangan pada tabel **orders**.

```python
missing_orders = order_items.loc[
    ~order_items["order_id"].isin(orders["order_id"])
]

print("=" * 70)
print("Relationship Check : Orders → Order Items")
print("=" * 70)

print(f"Total Order Items           : {len(order_items):,}")
print(f"Missing Orders              : {len(missing_orders):,}")

if len(missing_orders) == 0:
    print("Relationship Status         : PASSED ✅")
else:
    print("Relationship Status         : FAILED ❌")
```

### Orders → Payments

**Objective**

Memastikan seluruh transaksi pembayaran (`order_id`) memiliki pasangan pada tabel **orders**.

```python
missing_payments = payments.loc[
    ~payments["order_id"].isin(orders["order_id"])
]

print("=" * 70)
print("Relationship Check : Orders → Payments")
print("=" * 70)

print(f"Total Payments              : {len(payments):,}")
print(f"Missing Orders              : {len(missing_payments):,}")

if len(missing_payments) == 0:
    print("Relationship Status         : PASSED ✅")
else:
    print("Relationship Status         : FAILED ❌")
```

### Orders → Reviews

**Objective**

Memastikan seluruh ulasan (`order_id`) memiliki pasangan pada tabel **orders**.

```python
missing_reviews = reviews.loc[
    ~reviews["order_id"].isin(orders["order_id"])
]

print("=" * 70)
print("Relationship Check : Orders → Reviews")
print("=" * 70)

print(f"Total Reviews               : {len(reviews):,}")
print(f"Missing Orders              : {len(missing_reviews):,}")

if len(missing_reviews) == 0:
    print("Relationship Status         : PASSED ✅")
else:
    print("Relationship Status         : FAILED ❌")
```

### Order Items → Products

## Additional Data Quality Findings

Bagian ini merangkum temuan tambahan dari validasi cakupan temporal, konsistensi logika bisnis, dan nilai mustahil yang dilakukan lintas tabel selama proses profiling. Temuan ini menjadi acuan untuk keputusan cleaning pada tahap berikutnya dan akan dirujuk kembali pada `data_quality_report.md`.

### Identifier Consistency
- `customer_id` (Customers/Orders) bersifat unik per baris (identifier level order), sedangkan `customer_unique_id` TIDAK unik (96,096 unique dari 99,441 baris) — terdapat 3,345 baris dengan pelanggan yang order lebih dari sekali.
- **Keputusan yang perlu diambil:** seluruh analisis customer-level (segmentation, churn, RFM) wajib menggunakan `customer_unique_id`, bukan `customer_id`.

### Temporal Coverage
| Tabel | Kolom | Rentang |
|---|---|---|
| Orders | `order_purchase_timestamp` | 04-09-2016 → 17-10-2018 (~25 bulan) |
| Orders | `order_delivered_customer_date` | 11-10-2016 → 17-10-2018 |
| Reviews | `review_creation_date` | 02-10-2016 → 31-08-2018 |
| Reviews | `review_answer_timestamp` | 07-10-2016 → 29-10-2018 |

Catatan: total cakupan data ±25 bulan. Ini menjadi batas atas untuk validasi threshold churn 180 hari — periode observasi tidak cukup panjang untuk mengukur siklus churn di atas ~12–18 bulan dengan keyakinan tinggi.

### Business Logic Anomalies
| Tabel | Pemeriksaan | Jumlah Baris Anomali |
|---|---|---|
| Orders | `order_delivered_carrier_date` < `order_approved_at` | 1,359 |
| Orders | `order_delivered_customer_date` < `order_delivered_carrier_date` | 23 |
| Orders | status `delivered` tapi `order_delivered_customer_date` kosong | 8 |
| Reviews | `review_answer_timestamp` < `review_creation_date` | 0 |

**Keputusan yang perlu diambil:** 1,382 baris Orders (1,359 + 23) memiliki urutan tanggal yang tidak logis dan perlu ditinjau — apakah di-drop, dikoreksi, atau dibiarkan dengan flag khusus saat tahap cleaning nanti.

### Impossible Values
| Tabel | Pemeriksaan | Jumlah Baris Anomali |

| Order Items | `price` ≤ 0 | 0 |
| Order Items | `freight_value` < 0 | 0 |
| Payments | `payment_value` ≤ 0 | 9 |
| Payments | `payment_installments` ≤ 0 | 2 |
| Reviews | `review_score` di luar rentang 1–5 | 0 |

**Keputusan yang perlu diambil:** 11 baris Payments dengan nilai tidak valid perlu ditinjau — kemungkinan besar terkait transaksi yang dibatalkan/gratis, perlu dicek silang dengan `order_status` sebelum diputuskan di-drop atau tidak.

Skala ini tergolong kecil dan tidak menghambat kelayakan dataset secara keseluruhan, namun tetap wajib dicatat sebagai keputusan eksplisit di Decision Log sebelum data cleaning dimulai.

## Observation

Berdasarkan proses data profiling, seluruh sembilan dataset berhasil dimuat dan diperiksa tanpa kendala. Struktur dataset menunjukkan bahwa Olist merupakan dataset relasional yang dibangun menggunakan beberapa key utama seperti `order_id`, `product_id`, dan `seller_id`. Validasi relationship antar tabel menunjukkan bahwa foreign key utama berhasil dipetakan dengan benar (seluruh cross-table check PASSED), sehingga dataset memiliki integritas relasional yang baik.

Satu temuan penting: `customer_id` ternyata bersifat unik per baris (level order), sedangkan identifier pelanggan sesungguhnya adalah `customer_unique_id`, yang menunjukkan 3,345 baris pelanggan dengan lebih dari satu order. Temuan ini krusial karena seluruh analisis customer-level pada tahap berikutnya (segmentation, churn) wajib menggunakan `customer_unique_id`.

Validasi cakupan temporal menunjukkan data mencakup periode ±25 bulan (September 2016 – Oktober 2018), yang akan menjadi acuan dalam memvalidasi asumsi threshold churn 180 hari pada BRD. Validasi logika bisnis menemukan 1,382 baris pada tabel Orders dengan urutan tanggal yang tidak konsisten, serta 11 baris pada tabel Payments dengan nilai numerik yang tidak valid (payment_value atau installments ≤ 0) — totalnya sekitar 1.4% dari total order records. Kolom timestamp pada beberapa tabel juga masih bertipe `object` dan perlu dikonversi ke `datetime` pada tahap ETL.

Secara keseluruhan, tidak ditemukan permasalahan struktural yang menghambat kelayakan dataset.
